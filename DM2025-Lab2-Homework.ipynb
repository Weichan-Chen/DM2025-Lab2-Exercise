# **Competition Code**

## 1. Preprocessing Steps

```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import json
from pandas import json_normalize

#載入四個Dataset，並將final_posts.jason格式轉換為.csv(命名為df_fianl)

df_data_identification = pd.read_csv('/content/drive/MyDrive/Lab2_Phase3/data_identification.csv')
df_emotion = pd.read_csv('/content/drive/MyDrive/Lab2_Phase3/emotion.csv')
df_sampleSubmission = pd.read_csv('/content/drive/MyDrive/Lab2_Phase3/samplesubmission.csv')
df_final_data = pd.read_csv('/content/drive/MyDrive/Lab2_Phase3/final_data.csv')

path = '/content/drive/MyDrive/Lab2_Phase3/final_posts.json'
df = pd.read_json(path)

with open(path, 'r', encoding='utf-8') as f:data = json.load(f)
df_final = json_normalize(data, sep='_')
df_final.to_csv('/content/drive/MyDrive/Lab2_Phase3/final_data.csv',
          index=False, encoding='utf-8-sig')

#查看df的欄位
print(df_final.columns[:50])

#保留欄位
df = df_final[[
    'root__source_post_post_id',
    'root__source_post_text',
    'root__source_post_hashtags'
        ]].copy()

#更改欄位名稱
df = df.rename(columns={
    'root__source_post_post_id': 'id',
    'root__source_post_text': 'text',
    'root__source_post_hashtags': 'hashtags'
            })

#將final_post與emotion和data_identificaton合併

df = df.merge(df_emotion, on='id', how='left')
df = df.merge(df_data_identification, on='id', how='left')

empty_hashtags = df['hashtags'].apply(lambda x: isinstance(x, list) and len(x) == 0)
print("[],空列表數量:", empty_hashtags.sum())

# 看這些 [] 列的 emotion 分布
empty_emotion_counts = df.loc[empty_hashtags, 'emotion'].value_counts()
print("\nhashtag 為 [] 的列，其 emotion 分布：")
print(empty_emotion_counts)
print("\n佔所有訓練資料的比例：")
print(len(df))
#刪除欄位hashtags
df = df.drop(columns=['hashtags'])

檢查df的所有資料中，emotion為"NaN"的資料是否split也都是"test"
mask_nan = df['emotion'].isna()
all_ok = (df.loc[mask_nan, 'split'] == 'test').all()
print("所有 emotion 為 NaN 的列，split 是否都為 'test'？", all_ok)

#將df分為Train、Predict
df_train = df[df['split'] == 'train'].copy()
df_predict = df[df['split'] == 'test'].copy()

#刪除欄位split
df_train = df_train.drop(columns=['split'])
df_predict = df_predict.drop(columns=['split'])

#查看Train、Predict的欄位及筆數

print("df_train的shape：", df_train.shape)
print("df_train欄位：",df_train.columns)
print("---------------------------------------------------------------")
print("df_predict的shape：", df_predict.shape)
print("df_predict欄位：",df_predict.columns)

#檢查缺失值
print("---------------------------------------------------------------")
print("訓練資料的缺失值","\n", df_train.isnull().sum())
print("---------------------------------------------------------------")
print("預測資料的缺失值","\n", df_predict.isnull().sum())

#查看df_train的emotion distribution
print("---------------------------------------------------------------")
emotion_counts = df_train['emotion'].value_counts()
print("訓練資料emotion的分布",emotion_counts)
emotion_counts.plot(kind = "bar",
          title = "Emotion Distribution",
          ylim =[1000,25000],
          rot = 0, fontsize = 11, figsize = (8,3))
#將df_train切成訓練集與測試集
from sklearn.model_selection import train_test_split
train_df, test_df = train_test_split(
    df_train,
    test_size=0.3,
    random_state=42,
    stratify=df_train['emotion']
)
```

## 2. Feature Engineering Steps

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
TIF_500 = TfidfVectorizer(
    max_features=20000,
    min_df=5,
    max_df=0.9,
    ngram_range=(1, 2),
    sublinear_tf=True
)
TIF_500.fit(train_df["text"])

X_train = TIF_500.transform(train_df['text'])
y_train = train_df["emotion"]
X_test  = TIF_500.transform(test_df["text"])
y_test  = test_df['emotion']

# LabelEncoder
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_train_enc = le.fit_transform(y_train)
y_test_enc  = le.transform(y_test)

```

## 3. Model Implementation Steps

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, StratifiedKFold,RandomizedSearchCV
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
from sklearn.model_selection import RandomizedSearchCV

#建一條 Pipeline
pipe = Pipeline([
    ("sampler", RandomOverSampler(random_state=42)),
    ("clf", XGBClassifier(
        objective='multi:softmax',
        num_class=len(le.classes_),
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=2,
        reg_alpha=0.5,
        n_jobs=-1,
        eval_metric='mlogloss',
        tree_method='hist',
        random_state=42
    ))
])

#設定要「網格搜尋」的超參數
param_grid = {
    "clf__max_depth": [4, 5],
    "clf__n_estimators": [200, 350],
    "clf__learning_rate": [0.05],
}

#設定交叉驗證方式
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

#建立 GridSearchCV 物件
grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

#在「X_train, y_train_enc」上做網格搜尋 + 交叉驗證
grid.fit(X_train, y_train_enc)

print("\n最佳參數：", grid.best_params_)
print("CV 最佳 macro F1：", grid.best_score_)


import numpy as np
from scipy.sparse import vstack

#合併全部有標籤的資料
X_full = vstack([X_train, X_test])
y_full = np.concatenate([y_train_enc, y_test_enc], axis=0)

#取出GridSearch 找到的最佳參數
best_params = grid.best_params_

xgb_best_params = {k.replace("clf__", ""): v for k, v in best_params.items()}

from imblearn.pipeline import Pipeline
from imblearn.over_sampling import RandomOverSampler
from xgboost import XGBClassifier

final_model = Pipeline([
    ("sampler", RandomOverSampler(random_state=42)),
    ("clf", XGBClassifier(
        objective='multi:softmax',
        num_class=len(le.classes_),
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.5,
        reg_alpha=0.3,
        n_jobs=-1,
        eval_metric='mlogloss',
        tree_method='hist',
        random_state=42,
        **xgb_best_params
    ))
])

#用全部資料重訓
final_model.fit(X_full, y_full)


#預測測試資料並輸出成CSV檔
X_predict = TIF_500.transform(df_predict["text"])


y_predict_enc = final_model.predict(X_predict)


y_predict_labels = le.inverse_transform(y_predict_enc)


df_pred = pd.DataFrame({
    'id': df_predict['id'].values,
    'emotion': y_predict_labels
})

df_out = df_predict[['id']].merge(df_pred, on='id', how='left')
df_out.to_csv("result_final.csv", index=False)
print("已輸出 result_final_1.csv")
```
